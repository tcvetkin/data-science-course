```python
while True:
    text = input("Введите текст (минимум 20 символов): ")
    if len(text) >= 20:
        break
    else:
        print(f"Текст слишком короткий! Вы ввели {len(text)} символов. Попробуйте снова.\n")
print("\n" + "=" * 50)
print("РЕЗУЛЬТАТЫ АНАЛИЗА ТЕКСТА")
print("=" * 50)
print(f"1. Длина текста: {len(text)} символов")
print(f"2. Текст в верхнем регистре:\n{text.upper()}")
print(f"3. Текст в нижнем регистре:\n{text.lower()}")
words = text.split()
print(f"4. Количество слов: {len(words)}")
if words:
    print(f"5. Первое слово: \"{words[0]}\"")
    print(f"   Последнее слово: \"{words[-1]}\"")
python_found = "python" in text.lower()
print(f"\n6. Содержит ли текст слово \"Python\"? {'ДА' if python_found else 'НЕТ'}")
text_with_hyphens = text.replace(" ", "-")
print(f"\n7. Текст с заменами пробелов на дефисы:\n{text_with_hyphens}")
print("\n" + "=" * 50)
print("ДОПОЛНИТЕЛЬНАЯ ИНФОРМАЦИЯ:")
print(f"  - Количество символов без пробелов: {len(text.replace(' ', ''))}")
print(f"  - Количество пробелов: {text.count(' ')}")
print(f"  - Количество цифр в тексте: {sum(c.isdigit() for c in text)}")
print(f"  - Количество букв в тексте: {sum(c.isalpha() for c in text)}")
print("=" * 50)
```
